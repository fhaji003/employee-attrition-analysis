CLEANINING CODE (Code below):

```python
df = df.drop(columns=["EmployeeCount", "Over18", "StandardHours", "EmployeeNumber"])

df.shape
```

```python
df["AttritionFlag"] = df["Attrition"].map({"Yes": 1, "No": 0})

df["AttritionFlag"].mean()
```

Charts && 
Attrition Count:

```python
sns.countplot(data=df, x="Attrition")
plt.title("Employee Attrition Count")
plt.show()
```

```python
sns.barplot(data=df, x="OverTime", y="AttritionFlag")
plt.title("Attrition Rate by Overtime")
plt.ylabel("Attrition rate") 
plt.show()
```

```python
sns.barplot(data=df, y="JobRole", x="AttritionFlag",
            order=df.groupby("JobRole")["AttritionFlag"].mean().sort_values(ascending=False).index)
plt.title("Attrition Rate by Job Role")
plt.xlabel("Attrition rate")
plt.show()
```

```python
sns.boxplot(data=df, x="Attrition", y="MonthlyIncome")
plt.title("Monthly Income by Attrition")
plt.show()
```

```python
sns.histplot(data=df, x="Age", hue="Attrition")
plt.title("Employee Age Distribution by Attrition")
plt.show()
```
